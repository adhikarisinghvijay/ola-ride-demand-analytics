```python
#Column Season
print(df["Season"].value_counts())
print("_" * 80)
print(df["Season"].value_counts(normalize = True) * 100)
print("_" * 80)
print(df["Season"].nunique())
print("_" * 80)
print("\n")

df["Season"].value_counts().plot(kind = "bar")
plt.title("Bar graph of different seasons", fontsize = 20)
plt.xlabel("Season type")
plt.ylabel("Frequency")
plt.show()
```

```python
#Coulmn ["Season", "Weather"]
for season in range(1,df["Season"].nunique() + 1):
    season_weather = df[df["Season"] == season][["Season", "Weather"]]
    print(season_weather.value_counts())
    season_weather.value_counts().plot(kind = "bar")
    plt.title(f"Bar plot of season {season} with weather", fontsize = 20)
    plt.xlabel("Season & Weather")
    plt.ylabel("Frequency")
    plt.show()
    print("_" * 70)
```

```python
#Columns ["Season", "Count"]
for season in range(1, df["Season"].nunique() + 1):
    season_count = df[df["Season"]  == season][["Season", "Count"]]
    print(season_count.value_counts())
    season_count.value_counts().head(30).plot(kind = "bar")
    plt.title(f"Bar plot of season {season} with count", fontsize = 20)
    plt.xlabel("Season & Count")
    plt.ylabel("Frequency")
    plt.show()
```

```python
#Columns ["Weather", "Count"]
for season in range(1, df["Weather"].nunique() + 1):
    season_count = df[df["Weather"]  == season][["Weather", "Count"]]
    print(season_count.value_counts())
    season_count.value_counts().head(30).plot(kind = "bar")
    plt.title(f"Bar plot of weather {season} with count", fontsize = 20)
    plt.xlabel("Weather & Count")
    plt.ylabel("Frequency")
    plt.show()
```

```python
#Columns ["Casual", "Registered", "Count]
casual_count = df["Casual"].nunique()
registered_count = df["Registered"].nunique()
count_count = df["Count"].nunique()

print((casual_count / count_count) * 100)
print((registered_count / count_count) * 100)
```
